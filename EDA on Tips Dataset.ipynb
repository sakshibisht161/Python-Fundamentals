# Data Visualization

```python
df["day"].value_counts().plot(kind = "bar" , color ="darkorange")
```

# Countplot

```python
sns .countplot(x = "sex",data = df , color ="r")
```

```python
sns .countplot(x = "sex",data = df , hue = "sex")
```

# displot

```python
sns.displot(x = "time" , data = df , color = "teal")
```

# Histplot 

```python
sns.histplot(df["tip" ],kde = True , color ="m")
plt.show()
```

```python
sns.histplot(df["total_bill" ],kde = True ,color = "gold")
plt.show()
```

# Box plot

```python
sns.boxplot( x = "day" , y ="total_bill" , data = df , color ="c")
plt.show()
```

# Scatter plot 

```python
sns.scatterplot( x = "total_bill" , y ="tip" , data = df , hue = "time")
plt.show()
```

# heatmap 

```python
sns.heatmap(df.corr(numeric_only = True ),annot = True , cmap = "coolwarm")
plt.show()
```
