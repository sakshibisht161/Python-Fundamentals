# Data Visualization

# Bar Chart

```python
df["Gender"].value_counts().plot(kind = "bar" , color = "cyan")
```

# pie 

```python
df["Gender"].value_counts().plot(kind = "pie" , autopct = "%1.1f%%" , title = "Gender Distribution")
plt.ylabel(' ')
plt.show()
```

# Hist plot 

```python
sns.histplot(x = "Age",color ="teal" , data = df, kde = True)
```

# Countplot 

```python
sns.countplot(x = "Gender" , palette = "Set2" , data = df )
```

# Displot 

```python
sns.displot(x = "Gender" , color = "m" , data = df )
```

# Box plot

```python
sns.boxplot(x ="Gender" , y = "Annual Income (k$)" , hue = "Gender" , data = df)
```

# scatter chart 

```python
sns.scatterplot(x = "Annual Income (k$)", y = "Spending Score (1-100)", data = df ,hue = "Gender")
```

# Pair plot 

```python
sns.pairplot(df,hue = "Gender")
```
