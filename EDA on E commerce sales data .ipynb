# Data Visualization 

# Bar chart 

```python
df["Product line"].value_counts().plot(kind = "bar" , color = "g")
```

# Pie Chart 

```python
df["Gender"].value_counts().plot(kind = "pie" , autopct = "%1.1F%%" , title = "Gender distribution ")
plt.show()
```

# Displot 

```python
sns.displot(x ="Branch" , data = df , hue = "Branch")
```

# Countplot 

```python
sns.countplot(x ="City" , data = df , hue = "City")
```

# Box plot 

```python
sns.boxplot(x = "Customer type" ,y = "Sales" ,data = df , hue = "Gender")
```

# Histplot 

```python
sns.histplot(x = "Quantity" ,data = df ,kde = True ,color ="cyan")
plt.show()
```

```python
sns.histplot(x = "Rating" ,data = df ,kde = True ,color ="r")
plt.show()
```
