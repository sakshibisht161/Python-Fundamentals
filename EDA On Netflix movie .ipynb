# Data Visualization

```python
plt.figure(figsize=(8,5))
plt.hist(df['Popularity'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Movie Popularity')
plt.xlabel('Popularity')
plt.ylabel('Number of Movies')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.countplot(data=df, y='Genre', order=df['Genre'].value_counts().index[:10])
plt.title('Top 10 Movie Genres')
plt.show()
```

# What is the most frequent genre of movies released on Netflix?

```python
df["Genre"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Movie Genres")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.show()
```

# Which movie has the highest popularity? What is its genre?

```python
top = df[df["Popularity"] == df["Popularity"].max()]

print(top[["Title", "Genre"]])
```

```python
top
```

# Which movie has the lowest popularity? What is its genre?

```python
lower = df[df["Popularity"] == df["Popularity"].min()]

print(lower[["Title", "Genre"]])
```
